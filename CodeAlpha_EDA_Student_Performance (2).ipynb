
## 6. Hypothesis Testing

### Objective
To investigate whether average mathematics
scores differ between students who completed
test preparation and those who did not.

### Statistical Test
Independent two-sample Welch's t-test.

### Hypotheses

H0: The average mathematics scores of both
groups are equal.

H1: The average mathematics scores of both
groups are different.

### Significance Level
0.05

### Interpretation
The p-value is compared against the
significance level to determine whether
the difference is statistically significant.


```python
from scipy.stats import ttest_ind
```

```python
completed = df[
    df["test preparation course"] == "completed"
]["math score"]

not_completed = df[
    df["test preparation course"] == "none"
]["math score"]
```

```python
t_stat, p_value = ttest_ind(
    completed,
    not_completed,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("There is a statistically significant difference.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is insufficient evidence of a difference.")
```
